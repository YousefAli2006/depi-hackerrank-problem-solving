```python
# Hackerrank Prepare->Algorithms->Compare the Triplets
# (https://www.hackerrank.com/challenges/compare-the-triplets/problem?isFullScreen=true)
def compareTriplets(a,b):
    a_sum=0
    b_sum=0
    for i in range(3):
        if a[i]>b[i]:
            a_sum+=1
        elif a[i]<b[i]:
            b_sum+=1
    return [a_sum,b_sum]
a=list(map(int,input().split(" ")))
b=list(map(int,input().split(" ")))
print(*compareTriplets(a,b))
# input:
# 5 6 7
# 3 6 10
```

```python
# Hackerrank Prepare->Algorithms->Plus Minus
# (https://www.hackerrank.com/challenges/plus-minus/problem?isFullScreen=true)
def plusMinus(arr):
    positive_counter=0
    negative_counter=0
    zero_counter=0
    for i in arr:
        if i>0:
            positive_counter+=1
        elif i<0:
            negative_counter+=1
        else:
            zero_counter+=1
    return [positive_counter/len(arr),negative_counter/len(arr),zero_counter/len(arr)]
n=int(input())
arr=list(map(int,input().split(" ")))
print(*plusMinus(arr),sep="\n")
# input:
# 6
# -4 3 -9 0 4 1
```

```python
# Hackerrank Prepare->Algorithms->Mini-Max Sum
# (https://www.hackerrank.com/challenges/mini-max-sum/problem?isFullScreen=true)
def miniMaxSum(arr):
    min_sum=sum(arr)-max(arr)
    max_sum=sum(arr)-min(arr)
    return [min_sum,max_sum]
arr=list(map(int,input().split(" ")))
print(*miniMaxSum(arr))
# input:
# 1 2 3 4 5
```

```python
# Hackerrank Prepare->Algorithms->Subarray Division
# (https://www.hackerrank.com/challenges/the-birthday-bar/problem?isFullScreen=true)
def birthday(s,d,m):
    ways=[]
    for i in range(len(s)):
        if sum(s[i:i+m])==d:
            ways.append(s[i:i+m])
    return len(ways)
n=int(input())
s=list(map(int,input().split(" ")))
d,m=list(map(int,input().split(" ")))
print(birthday(s,d,m))
# input:
# 5
# 1 2 1 3 2
# 3 2
```

```python
# Hackerrank Prepare->Algorithms->Divisible Sum Pairs
# (https://www.hackerrank.com/challenges/divisible-sum-pairs/problem?isFullScreen=true)
def divisibleSumPairs(n,k,ar):
    pairs=[]
    for i in range(len(ar)):
        for j in range(len(ar)):
            if i<j and (ar[i]+ar[j])%k==0:
                pairs.append((i,j))
    return len(pairs)
n,k=list(map(int,input().split(" ")))
ar=list(map(int,input().split(" ")))
print(divisibleSumPairs(n,k,ar))
#input:
# 6 3
# 1 3 2 6 1 2 
```

```python
# Hackerrank Prepare->Algorithms->Electronics Shop
# (https://www.hackerrank.com/challenges/electronics-shop/problem?isFullScreen=true)
def getMoneySpent(keyboards,drives,b):
    x=[k+d for k in keyboards for d in drives if b-(k+d)>=0]
    x.append(-1)
    return max(x)
b,n,m=list(map(int,input().split(" ")))
keyboards=list(map(int,input().split(" ")))
drives=list(map(int,input().split(" ")))
print(getMoneySpent(keyboards,drives,b))
#input:
# 10 2 3
# 3 1
# 5 2 8
```

```python
# Hackerrank Prepare->Algorithms->Beautiful Days at the Movies
# (https://www.hackerrank.com/challenges/beautiful-days-at-the-movies/problem?isFullScreen=true)
def beautifulDays(i,j,k):
    nums=list(range(i,j+1))
    nums_reverse=map(lambda x:int(str(x)[::-1]),list(range(i,j+1)))
    return sum(list(map(lambda a,b:abs(a-b)%k==0,nums,nums_reverse)))
i,j,k=map(int,input().split(" "))
print(beautifulDays(i,j,k))
#input:
# 20 23 6
```

```python
# Hackerrank Prepare->Algorithms->Viral Advertising
# (https://www.hackerrank.com/challenges/strange-advertising/problem?isFullScreen=true)
def viralAdvertising(n):
    likers=0
    cumulative_likers=0
    receivers=5
    for _ in range(n):
        likers=receivers//2
        cumulative_likers+=likers
        receivers=likers*3
    return cumulative_likers
n=int(input())
print(viralAdvertising(n))
# input:
# 3
```

```python
# Hackerrank Prepare->Algorithms->Jumping on the Clouds: Revisited
# (https://www.hackerrank.com/challenges/jumping-on-the-clouds-revisited/problem?isFullScreen=true)
def jumpingOnClouds(c,k):
    e=100
    cloud_index=0
    while True:
        cloud_index=(cloud_index+k)%len(c)
        if c[cloud_index]==0:
            e-=1
        else:
            e-=3
        if cloud_index==0:
            break
    return e
n,k=map(int,input().split(" "))
c=list(map(int,input().split(" ")))
print(jumpingOnClouds(c,k))
# input:
# 8 2
# 0 0 1 0 0 1 1 0
```

```python
# Hackerrank Prepare->Algorithms->Utopian Tree
# (https://www.hackerrank.com/challenges/utopian-tree/problem?isFullScreen=true)
def utopianTree(n):
    h=1
    for i in range(1,n+1):
        if i%2!=0:
            h*=2
        else:
            h+=1
    return h
t=int(input())
for _ in range(t):
    print(utopianTree(int(input())))
#input:
# 3
# 0
# 1
# 4
```

```python
# Hackerrank Prepare->Algorithms->Circular Array Rotation
# (https://www.hackerrank.com/challenges/circular-array-rotation/problem?isFullScreen=true)
def circularArrayRotation(a,k,queries):
    return [a[(q-k%len(a))%len(a)] for q in queries]
n,k,q=map(int,input().split(" "))
a=list(map(int,input().split(" ")))
queries=[int(input()) for _ in range(q)]
print(*circularArrayRotation(a,k,queries),sep="\n")
# input:
# 3 2 3
# 1 2 3
# 0
# 1
# 2
```

```python
# Hackerrank Prepare->Algorithms->Fibonacci Modified
# (https://www.hackerrank.com/challenges/fibonacci-modified/problem?isFullScreen=true)
import sys
sys.set_int_max_str_digits(0)
def fibonacciModified(t1,t2,n):
    if n==1:
        return t1
    if n==2:
        return t2
    for _ in range(n-2):
        t1,t2=t2,t1+t2**2
    return t2
t1,t2,n=map(int,input().split(" "))
print(fibonacciModified(t1,t2,n))
#input:
# 0 1 5
```

```python
# Hackerrank Prepare->Algorithms->Angry Professor
# (https://www.hackerrank.com/challenges/angry-professor/problem?isFullScreen=true)
def angryProfessor(k,a):
    on_time = sum([1 for t in a if t<=0])
    return "YES" if on_time<k else "NO"
t=int(input())
for _ in range(t):
    n,k=map(int,input().split())
    a=list(map(int,input().split()))
    print(angryProfessor(k,a))
# input:
# 2
# 4 3
# -1 -3 4 2
# 4 2
# 0 -1 2 1
```

```python
# Hackerrank Prepare->Algorithms->The Coin Change Problem
# (https://www.hackerrank.com/challenges/coin-change/problem?isFullScreen=true)
def getWays(n,c):
    dp=[0]*(n+1)
    dp[0]=1
    for coin in c:
        for i in range(coin,n+1):
            dp[i]+=dp[i-coin]
    return dp[n]
n,m=map(int,input().split())
c=list(map(int,input().split()))
print(getWays(n,c))
# input:
# 10 4
# 2 5 3 6
```

```python
# Hackerrank Prepare->Algorithms->Between Two Sets
# (https://www.hackerrank.com/challenges/between-two-sets/problem?isFullScreen=true)
from math import gcd
def getTotalX(a,b):
    my_lcm=a[0]
    for x in a[1:]:
        my_lcm=my_lcm*x//gcd(my_lcm,x)
    my_gcd=b[0]
    for x in b[1:]:
        my_gcd=gcd(my_gcd,x)
    c=0
    for i in range(my_lcm,my_gcd+1,my_lcm):
        if my_gcd%i==0:
            c+=1
    return c

n,m=map(int,input().split())
a=list(map(int,input().split()))
b=list(map(int,input().split()))
print(getTotalX(a,b))
# input:
# 2 3
# 2 4
# 16 32 96
```
