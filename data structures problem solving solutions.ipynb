```python
# Hackerrank Prepare->Data Structures->Reverse a linked list
# (https://www.hackerrank.com/challenges/reverse-a-linked-list/problem?isFullScreen=true)

import math
import os
import random
import re
import sys

os.environ['OUTPUT_PATH']='output.txt' #additional line for making undeletable hackerrank code work here

class SinglyLinkedListNode:
    def __init__(self, node_data):
        self.data = node_data
        self.next = None

class SinglyLinkedList:
    def __init__(self):
        self.head = None
        self.tail = None

    def insert_node(self, node_data):
        node = SinglyLinkedListNode(node_data)

        if not self.head:
            self.head = node
        else:
            self.tail.next = node


        self.tail = node

def print_singly_linked_list(node, sep, fptr):
    while node:
        fptr.write(str(node.data))

        node = node.next

        if node:
            fptr.write(sep)
# my code:
def reverse(head):
    if head.next is None:
        return head
    my_stack=[]
    current=head
    while current is not None:
        my_stack.append(current.data)
        current=current.next
    my_stack.reverse()
    llist_reversed=SinglyLinkedList()
    for x in my_stack:
        llist_reversed.insert_node(x)
    return llist_reversed.head
#end of my code
if __name__ == '__main__':
    fptr = open(os.environ['OUTPUT_PATH'], 'w')

    tests = int(input())

    for tests_itr in range(tests):
        llist_count = int(input())

        llist = SinglyLinkedList()

        for _ in range(llist_count):
            llist_item = int(input())
            llist.insert_node(llist_item)

        llist1 = reverse(llist.head)

        print_singly_linked_list(llist1, ' ', fptr)
        fptr.write('\n')

    fptr.close()

#additional code to print output into the cell output and delete extra file:
output_path = os.environ.get('OUTPUT_PATH')
with open(output_path,'r') as f:
    print(f.read())
os.remove(output_path)

#input:
# 1
# 5
# 1
# 2
# 3
# 4
# 5
```

```python
# Hackerrank Prepare->Data Structures->Array Manipulation
# (https://www.hackerrank.com/challenges/crush/problem?isFullScreen=true)
def arrayManipulation(n,queries):
    diff=[0]*(n+2)
    for a,b,k in queries:
        diff[a]+=k
        diff[b+1]-=k
    running=0
    max_val=0
    for i in range(1,n+1):
        running+=diff[i]
        if running>max_val:
            max_val=running
    return max_val
n,q=[int(x) for x in input().split(" ")]
queries=[[int(x) for x in input().split(" ")] for _ in range(q)]
print(arrayManipulation(n,queries))

#input:
# 5 3
# 1 2 100
# 2 5 100
# 3 4 100
```

```python
# Hackerrank Prepare->Data Structures->Dynamic Array
# (https://www.hackerrank.com/challenges/dynamic-array/problem?isFullScreen=true)
def dynamicArray(n,queries):
    seq_list = [[] for _ in range(n)]
    last_answer = 0
    answers = []
    for query in queries:
        q_type = query[0]
        x = query[1]
        y = query[2]
        idx=(x^last_answer)%n
        if q_type==1:
            seq_list[idx].append(y)
        else:
            seq=seq_list[idx]
            val=seq[y%len(seq)]
            last_answer=val
            answers.append(val)
    return answers
n,q=[int(x) for x in input().split(" ")]
queries=[[int(x) for x in input().split(" ")] for _ in range(q)]
print(*dynamicArray(n,queries),sep="\n")

#input:
# 2 5
# 1 0 5
# 1 1 7
# 1 0 3
# 2 1 0
# 2 1 1
```

```python
# Hackerrank Prepare->Data Structures->Queue using Two Stacks
# (https://www.hackerrank.com/challenges/queue-using-two-stacks/problem?isFullScreen=true)
class Queue:
    def __init__(self):
        self.stack_in=[]
        self.stack_out=[]
    def enqueue(self,x):
        self.stack_in.append(x)
    def dequeue(self):
        if not self.stack_out:
            while self.stack_in:
                self.stack_out.append(self.stack_in.pop())
        self.stack_out.pop()
    def peek(self):
        if not self.stack_out:
            while self.stack_in:
                self.stack_out.append(self.stack_in.pop())
        return self.stack_out[-1]

q=int(input())
queue=Queue()
for _ in range(q):
    query=input().split()
    if query[0]=='1':
        queue.enqueue(int(query[1]))
    elif query[0]=='2':
        queue.dequeue()
    elif query[0]=='3':
        print(queue.peek())
#input:
# 10
# 1 42
# 2
# 1 14
# 3
# 1 28
# 3
# 1 60
# 1 78
# 2
# 2
```

```python
# Hackerrank Prepare->Data Structures->Tree: Preorder Traversal
# (https://www.hackerrank.com/challenges/tree-preorder-traversal/problem?isFullScreen=true)
class Node:
    def __init__(self, info): 
        self.info = info  
        self.left = None  
        self.right = None 
        self.level = None 

    def __str__(self):
        return str(self.info) 

class BinarySearchTree:
    def __init__(self): 
        self.root = None

    def create(self, val):  
        if self.root == None:
            self.root = Node(val)
        else:
            current = self.root
         
            while True:
                if val < current.info:
                    if current.left:
                        current = current.left
                    else:
                        current.left = Node(val)
                        break
                elif val > current.info:
                    if current.right:
                        current = current.right
                    else:
                        current.right = Node(val)
                        break
                else:
                    break
#my code:
def preOrder(root):
    def helper_preOrder(node,result):
        if node:
            result.append(node)
            helper_preOrder(node.left,result)
            helper_preOrder(node.right,result)
        return result
    result=[]
    helper_preOrder(root,result)
    print(*result)
#end of my code
tree = BinarySearchTree()
t = int(input())

arr = list(map(int, input().split()))

for i in range(t):
    tree.create(arr[i])

preOrder(tree.root)
# input:
# 6
# 1 2 5 3 6 4
```

```python
# Hackerrank Prepare->Data Structures->Tree: Level Order Traversal
# (https://www.hackerrank.com/challenges/tree-level-order-traversal/problem?isFullScreen=true)
class Node:
    def __init__(self, info): 
        self.info = info  
        self.left = None  
        self.right = None 
        self.level = None 

    def __str__(self):
        return str(self.info) 

class BinarySearchTree:
    def __init__(self): 
        self.root = None

    def create(self, val):  
        if self.root == None:
            self.root = Node(val)
        else:
            current = self.root
         
            while True:
                if val < current.info:
                    if current.left:
                        current = current.left
                    else:
                        current.left = Node(val)
                        break
                elif val > current.info:
                    if current.right:
                        current = current.right
                    else:
                        current.right = Node(val)
                        break
                else:
                    break
#my code:
def levelOrder(root):
    if not root:
        return
    my_queue=[root]
    result=[]
    while my_queue:
        x=my_queue.pop(0)
        result.append(x.info)
        if x.left:
            my_queue.append(x.left)
        if x.right:
            my_queue.append(x.right)
    print(*result)
#end of my code
tree = BinarySearchTree()
t = int(input())

arr = list(map(int, input().split()))

for i in range(t):
    tree.create(arr[i])

levelOrder(tree.root)
#input:
# 6
# 1 2 5 3 6 4
```

```python
# Hackerrank Prepare->Data Structures->QHEAP1
# (https://www.hackerrank.com/challenges/qheap1/problem?isFullScreen=true)
class MinHeap:
    def __init__(self,initial=None):
        self.heap=[]
        if initial:
            self.heap=list(initial)
            self.heapify()
    def push(self,item):
        self.heap.append(item)
        self.sift_up(len(self.heap)-1)
    def pop(self):
        if not self.heap:
            raise IndexError("pop from empty heap")
        self.my_swap(0,-1)
        smallest=self.heap.pop()
        if self.heap:
            self.sift_down(0)
        return smallest
    def peek(self):
        if not self.heap:
            raise IndexError("peek from empty heap")
        return self.heap[0]
    def __len__(self):
        return len(self.heap)
    def heapify(self):
        for i in range(len(self.heap)//2-1,-1,-1):
            self.sift_down(i)
    def sift_up(self,idx):
        parent=(idx-1)//2
        while idx>0 and self.heap[idx]<self.heap[parent]:
            self.my_swap(idx,parent)
            idx=parent
            parent=(idx-1)//2
    def sift_down(self,idx):
        n=len(self.heap)
        while True:
            left=2*idx+1
            right=2*idx+2
            smallest=idx
            if left<n and self.heap[left]<self.heap[smallest]:
                smallest=left
            if right<n and self.heap[right]<self.heap[smallest]:
                smallest=right
            if smallest==idx:
                break
            self.my_swap(idx,smallest)
            idx=smallest
    def my_swap(self,i,j):
        self.heap[i],self.heap[j]=self.heap[j],self.heap[i]

q=int(input())
heap=MinHeap()
cnt={}
out=[]
for _ in range(q):
    p=input().split()
    t=int(p[0])
    if t==1:
        x=int(p[1])
        heap.push(x)
        cnt[x]=cnt.get(x,0)+1
    elif t==2:
        x=int(p[1])
        cnt[x]=cnt.get(x,0)-1
    else:
        while heap and cnt.get(heap.peek(),0)==0:
            heap.pop()
        out.append(str(heap.peek()))
print(*out,sep="\n")

#input:
# 5
# 1 4
# 1 9
# 3
# 2 4
# 3
```

```python
# Hackerrank Prepare->Data Structures->Equal Stacks
# (https://www.hackerrank.com/challenges/equal-stacks/problem?isFullScreen=true)
def equalStacks(h1,h2,h3):
    s1,s2,s3=sum(h1),sum(h2),sum(h3)
    i=j=k=0
    while True:
        if s1==s2==s3:
            return s1
        if i==len(h1)or j==len(h2)or k==len(h3):
            return 0
        if s1>=s2 and s1>=s3:
            s1-=h1[i]
            i+=1
        elif s2>=s1 and s2>=s3:
            s2-=h2[j]
            j+=1
        else:
            s3-=h3[k]
            k+=1
n1,n2,n3=map(int,input().split())
h1=list(map(int,input().split()))
h2=list(map(int,input().split()))
h3=list(map(int,input().split()))
print(equalStacks(h1,h2,h3))

#input:
# 5 3 4
# 3 2 1 1 1
# 4 3 2
# 1 1 4 1
```

```python
# Hackerrank Prepare->Data Structures->Jesse and Cookies
# (https://www.hackerrank.com/challenges/jesse-and-cookies/problem?isFullScreen=true)
import heapq
def cookies(k,arr):
    heapq.heapify(arr)
    n_operations=0
    while len(arr)>=2 and arr[0]<k:
        x=heapq.heappop(arr)
        y=heapq.heappop(arr)
        heapq.heappush(arr,x+2*y)
        n_operations+=1
    return n_operations if arr[0]>=k else -1
n,k=map(int,input().split())
arr=list(map(int,input().split()))
print(cookies(k,arr))

#input:
# 6 7
# 1 2 3 9 10 12
```

```python
# Hackerrank Prepare->Data Structures->Balanced Brackets
# (https://www.hackerrank.com/challenges/balanced-brackets/problem?isFullScreen=true)
def isBalanced(s):
    stack=[]
    pairs={')':'(',']':'[','}':'{'}
    for c in s:
        if c in pairs:
            if not stack or stack[-1]!=pairs[c]:
                return False
            stack.pop()
        else:
            stack.append(c)
    return not stack
n=int(input())
for _ in range(n):
    print("YES" if isBalanced(input()) else "NO")
#input:
# 3
# {[()]}
# {[(])}
# {{[[(())]]}} 
```

```python
# Hackerrank Prepare->Data Structures->Swap Nodes [Algo]
# (https://www.hackerrank.com/challenges/swap-nodes-algo/problem?isFullScreen=true)
import sys
sys.setrecursionlimit(2000)

class Node:
    def __init__(self,val):
        self.val=val
        self.left=None
        self.right=None

class BinaryTree:
    def __init__(self,L,R):
        self.L=L
        self.R=R
        self.root=self.constructor_helper(1,1)

    def constructor_helper(self,i,d):
        if i==-1:
            return None
        node=Node(i)
        node.depth=d
        node.left=self.constructor_helper(self.L[i],d+1)
        node.right=self.constructor_helper(self.R[i],d+1)
        return node

    def swap(self,k):
        def helper_swap(node):
            if not node:
                return
            if node.depth%k==0:
                node.left,node.right=node.right,node.left
            helper_swap(node.left)
            helper_swap(node.right)
        helper_swap(self.root)

    def inorder(self):
        res=[]
        def helper_inorder(node):
            if not node:
                return
            helper_inorder(node.left)
            res.append(node.val)
            helper_inorder(node.right)
        helper_inorder(self.root)
        return res

n=int(input())
L=[0]*(n+1)
R=[0]*(n+1)
for i in range(1,n+1):
    l,r=map(int,input().split())
    L[i]=l
    R[i]=r

tree=BinaryTree(L,R)
q=int(input())
for _ in range(q):
    k=int(input())
    tree.swap(k)
    print(*tree.inorder())
#input:
# 3
# 2 3
# -1 -1
# -1 -1
# 2
# 1
# 1
```
